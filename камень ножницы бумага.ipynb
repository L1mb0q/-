```python
import random
player_score = 0
pc_score = 0
while True:
  computer = random.randint(1, 3)
  player = int(input("Введите свой вариант(1 - stone 2 noz 3 paper): "))
  if computer == player:
    print("Ничья")
  elif (player == 1 and computer == 2) or (player == 2 and computer == 3) or (player == 3 and computer == 1):
    player_score += 1
    print("Победа")
  else:
    pc_score += 1
    print("Поражение")
    if player_score == 3 or pc_score == 3:
      print("Ты победил") if player_score == 3 else print("Ты проиграл машинам! Ехехехехех")
      cont = input("Do you wont to continue (yes/no)? ")
      if cont == "no":
       print("Пока пока")
       break
    else:
       pc_score = 0
       player_score = 0
```
