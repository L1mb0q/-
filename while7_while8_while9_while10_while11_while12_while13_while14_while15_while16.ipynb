```python
#While7
n = int(input("Введите n(>0): "))
k = 1
while k * k <= n:
    k += 1
print(f"Наименьшее k = {k}")
```

```python
#While8
n = int(input("Введите n(>0): "))
k = 1
while (k + 1) * (k + 1) <= n:
  k += 1
print(f"Наибольшее k = {k}")
```

```python
#While9
n = int(input("Введите n(>1): "))
k = 1
i = 3
while i <= n:
    i *= 3
    k += 1
print(f"Наименьшее k: {k}")
```

```python
#while13
a = float(input("Введите a: "))
k = 0
i = 0
while i <= a:
    k += 1
    i += 1 / k
print(f"Наименьшее k: {k}")
print(f"Сумма: {i}")
```

```python
#while14
a = float(input("Введите a: "))
k = 0
i = 0.0
while i + (1 / (k + 1)) < a:
    k += 1
    i += 1 / k
print(f"Наибольшее k: {k}")
print(f"Сумма: {i}")
```

```python
#while16
p = float(input("Введите P (0 < P < 50): "))
daily_distance = 10
total_distance = 0
days = 0
while total_distance <= 200:
    total_distance += daily_distance
    days += 1
    daily_distance += daily_distance * (p / 100)
print(f"Количество дней K: {days}")
print(f"Суммарный пробег S: {total_distance:.2f}")
```
