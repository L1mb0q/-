```python
#if16
A = float(input("Введите число A: "))
B = float(input("Введите число B: "))
C = float(input("Введите число C: "))
if A < B < C:
    A *= 2
    B *= 2
    C *= 2
else:
    A = -A
    B = -B
    C = -C

print(f"Новые значения: A = {A}, B = {B}, C = {C}")
```

```python
#if17
A = float(input("Введите число A: "))
B = float(input("Введите число B: "))
C = float(input("Введите число C: "))
if A < B < C or A > B > C:
    A *= 2
    B *= 2
    C *= 2
else:
    A = -A
    B = -B
    C = -C

print(f"Новые значения: A = {A}, B = {B}, C = {C}")
```

```python
#if19
a = int(input("Введите 1-е число: "))
b = int(input("Введите 2-е число: "))
c = int(input("Введите 3-е число: "))
d = int(input("Введите 4-е число: "))
if a == b == c:
    print("Порядковый номер отличного числа: 4")
elif a == b == d:
    print("Порядковый номер отличного числа: 3")
elif a == c == d:
    print("Порядковый номер отличного числа: 2")
else:
    print("Порядковый номер отличного числа: 1")
```

```python
#if20
A = float(input("Введите координату A: "))
B = float(input("Введите координату B: "))
C = float(input("Введите координату C: "))
dist_B = abs(B - A)
dist_C = abs(C - A)
if dist_B < dist_C:
    print(f"Точка B ближе к A. Расстояние: {dist_B}")
elif dist_C < dist_B:
    print(f"Точка C ближе к A. Расстояние: {dist_C}")
else:
    print("Точки B и C находятся на одинаковом расстоянии от A.")
```
